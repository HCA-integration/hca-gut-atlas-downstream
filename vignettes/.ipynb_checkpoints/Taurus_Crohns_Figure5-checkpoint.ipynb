```python
import scanpy as sc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python

path = "/Users/kylekimler/Projects/GCA/reference_mapping_benchmark/results/disease_mapping/TAURUS_raw_counts_annotated_final_annotated.h5ad"
adata = sc.read_h5ad(path)
```

```python

col_u = "uncertainty_Prelim_annotation"
col_y = "predicted_Prelim_annotation"
col_d = "Disease"

need = [col_u, col_y, col_d]
missing = [c for c in need if c not in adata.obs.columns]
if missing:
    raise ValueError("Missing obs columns: " + ", ".join(missing))

df = adata.obs[[col_d, col_y, col_u]].copy()
df = df.dropna(subset=[col_d, col_y, col_u])

disease_order = ["Healthy", "CD", "UC"]
df[col_d] = pd.Categorical(df[col_d], categories=disease_order, ordered=True)

# Helmsley GCA palette sampled from the GCA logo (navy, blue, cyan, green family)
gca_palette = {
    "navy": "#1f3c73",
    "blue": "#0c5ca3",
    "deep": "#0e385e",
    "cyan": "#0cc6de",
    "green": "#3dd18e",
    "teal": "#0893b4",
    "steel": "#095e89",
    "mint": "#31b182",
}

disease_colors = {
    "Healthy": gca_palette["green"],
    "CD": gca_palette["navy"],
    "UC": gca_palette["cyan"],
}

print(df[col_d].value_counts())

# Summary stats of uncertainty stratified by Disease and transferred label
summary = (
    df.groupby([col_d, col_y])[col_u]
    .agg(
        n="size",
        mean="mean",
        median="median",
        std="std",
        p10=lambda x: x.quantile(0.10),
        p25=lambda x: x.quantile(0.25),
        p75=lambda x: x.quantile(0.75),
        p90=lambda x: x.quantile(0.90),
    )
    .reset_index()
    .sort_values([col_d, "n"], ascending=[True, False])
)

out_csv = "entropy_stratified_by_disease_and_label.csv"
summary.to_csv(out_csv, index=False)
print("Wrote:", out_csv)

# Heatmap: mean uncertainty per predicted label x Disease (top labels by total n)
top_k = 40
top_labels = (
    summary.groupby(col_y)["n"].sum().sort_values(ascending=False).head(top_k).index
)
hm = (
    summary[summary[col_y].isin(top_labels)]
    .pivot(index=col_y, columns=col_d, values="mean")
    .reindex(index=top_labels)
    .reindex(columns=disease_order)
)

plt.figure(figsize=(7, 0.28 * len(hm) + 2))
plt.imshow(hm.values, aspect="auto")
plt.xticks(range(len(hm.columns)), hm.columns, rotation=0)
plt.yticks(range(len(hm.index)), hm.index)
plt.colorbar(label="Mean uncertainty")
plt.title("Mean uncertainty by transferred label and Disease")
plt.tight_layout()
plt.show()

# Boxplots per Disease for top labels within that Disease
def plot_boxes_for_disease(disease, top_k_per_disease=25):
    sub = df[df[col_d] == disease].copy()
    top = sub[col_y].value_counts().head(top_k_per_disease).index
    sub = sub[sub[col_y].isin(top)]
    labels = list(top)
    data = [sub.loc[sub[col_y] == lab, col_u].to_numpy() for lab in labels]

    plt.figure(figsize=(10, 4))
    bp = plt.boxplot(
        data,
        labels=labels,
        showfliers=False,
        patch_artist=True,
    )
    face = disease_colors.get(disease, gca_palette["blue"])
    for b in bp["boxes"]:
        b.set_facecolor(face)
        b.set_alpha(0.35)
    for w in bp["whiskers"]:
        w.set_color(face)
    for c in bp["caps"]:
        c.set_color(face)
    for m in bp["medians"]:
        m.set_color(gca_palette["deep"])

    plt.xticks(rotation=90)
    plt.ylabel("Uncertainty")
    plt.title("Uncertainty by transferred label in " + str(disease))
    plt.tight_layout()
    plt.show()

for d in disease_order:
    plot_boxes_for_disease(d, top_k_per_disease=25)

# Optional: quantile binning of uncertainty within each Disease
df["uncertainty_bin"] = (
    df.groupby(col_d)[col_u]
    .transform(lambda x: pd.qcut(x, q=4, labels=["Q1", "Q2", "Q3", "Q4"]))
)
bin_ct = pd.crosstab(df[col_d], df["uncertainty_bin"], normalize="index")
print(bin_ct)
```
