# Program Scores Across Gut Segments

This notebook analyzes gene program scores across different gut segments (tissues) in the healthy gut atlas data.

## Overview
- Load AnnData object with all cell type lineages
- Filter to specific gut segments
- Compute program scores for user-defined gene programs
- Aggregate scores by cell type and tissue
- Test for segment effects within cell types
- Visualize program scores across segments
- Analyze tissue specificity of programs

## Compute Program Scores

```python
# Check var_names format (they should be Ensembl IDs)
print("First 10 var_names:", adata.var_names[:10].tolist())
print("var_names are Ensembl IDs - will use gene_symbol_to_ensg mapping to convert gene symbols")
```

```python
# Function to map gene symbols to var_names (Ensembl IDs)
def get_gene_names(genes, adata):
    """
    Map gene symbols to var_names (Ensembl IDs) in adata.
    Returns var_names (Ensembl IDs) that can be used with scanpy.
    
    Parameters:
    -----------
    genes : list
        List of gene symbols (or Ensembl IDs)
    adata : AnnData
        AnnData object with var_names as Ensembl IDs
    
    Returns:
    --------
    found_genes : list
        List of var_names (Ensembl IDs) found in adata
    missing_genes : list
        List of genes that could not be mapped or found
    """
    var_name_set = set(adata.var_names)
    found_genes = []  # Will contain var_names (Ensembl IDs)
    missing_genes = []
    
    for gene in genes:
        gene_upper = str(gene).upper()
        
        # First, try direct match with var_names (in case user provided Ensembl IDs)
        if gene_upper in var_name_set:
            found_genes.append(gene_upper)
        # Otherwise, try mapping from gene symbol to Ensembl ID
        elif gene_upper in gene_symbol_to_ensg:
            ensg_id = gene_symbol_to_ensg[gene_upper]
            # Check if this Ensembl ID exists in adata
            if ensg_id in var_name_set:
                found_genes.append(ensg_id)
            else:
                missing_genes.append(gene)
        else:
            missing_genes.append(gene)
    
    return found_genes, missing_genes

# Compute scores for each program
program_scores = {}
for prog_name, genes in programs.items():
    print(f"\nComputing score for: {prog_name}")
    gene_names, missing = get_gene_names(genes, adata)
    
    if len(missing) > 0:
        print(f"  Warning: {len(missing)} genes not found: {missing[:5]}...")
    
    if len(gene_names) == 0:
        print(f"  Error: No genes found for {prog_name}, skipping...")
        continue
    
    # Compute score using scanpy
    # Note: use_raw=True if you want to use raw counts, use_raw=False for normalized data
    # Check if raw data is available
    use_raw = adata.raw is not None
    sc.tl.score_genes(adata, gene_names, score_name=prog_name, use_raw=use_raw)
    program_scores[prog_name] = adata.obs[prog_name].values
    print(f"  Computed score for {len(gene_names)} genes")

print(f"\nComputed scores for {len(program_scores)} programs")
```

## Aggregate Program Scores

Aggregate mean program scores by (cell type, tissue) and by (cell type, tissue, assay).

```python
# Create aggregation dataframes
agg_data = []

for prog_name in program_scores.keys():
    # Aggregate by (Prelim annotation, tissue)
    agg_ct_tissue = adata.obs.groupby(['Prelim annotation', 'tissue'])[prog_name].agg(['mean', 'std', 'count']).reset_index()
    agg_ct_tissue['program'] = prog_name
    agg_ct_tissue['aggregation_level'] = 'celltype_tissue'
    agg_data.append(agg_ct_tissue)
    
    # Aggregate by (Prelim annotation, tissue, assay)
    agg_ct_tissue_assay = adata.obs.groupby(['Prelim annotation', 'tissue', 'assay'])[prog_name].agg(['mean', 'std', 'count']).reset_index()
    agg_ct_tissue_assay['program'] = prog_name
    agg_ct_tissue_assay['aggregation_level'] = 'celltype_tissue_assay'
    agg_data.append(agg_ct_tissue_assay)

# Combine all aggregations
agg_df = pd.concat(agg_data, ignore_index=True)
agg_df = agg_df.rename(columns={'mean': 'mean_score', 'std': 'std_score', 'count': 'n_cells'})

print(f"Aggregated data shape: {agg_df.shape}")
print(f"\nAggregation levels: {agg_df['aggregation_level'].value_counts()}")
print(f"\nSample of aggregated data:")
print(agg_df.head(10))
```

```python
# Save aggregated data
agg_df.to_csv(output_dir / 'aggregated_program_scores.csv', index=False)
print(f"Saved aggregated scores to: {output_dir / 'aggregated_program_scores.csv'}")
```

## Test for Segment Effects

Test for segment effects within each cell type using an ordered tissue axis. We'll use a simple linear model approach.

```python
# Define tissue order along the gut axis
tissue_order = [
    "duodenum",
    "jejunum", 
    "ileum",
    "vermiform_appendix",
    "small_intestine",
    "ascending_colon",
    "descending_colon",
    "sigmoid_colon",
    "colon",
    "rectum",
    "gastrointestinal_system_mesentery"
]

# Create tissue order mapping
tissue_to_order = {tissue: idx for idx, tissue in enumerate(tissue_order)}

# Add tissue order to aggregated data
agg_df['tissue_order'] = agg_df['tissue'].map(tissue_to_order)
# Handle any tissues not in the order (set to NaN for now)
agg_df['tissue_order'] = agg_df['tissue_order'].fillna(999)

print("Tissue order mapping:")
for tissue, order in sorted(tissue_to_order.items(), key=lambda x: x[1]):
    print(f"  {order}: {tissue}")
```

```python
from sklearn.linear_model import LinearRegression

# Function to test segment effect for a single program and cell type
def test_segment_effect(agg_subset, covariates=None):
    """
    Test for segment effect using linear regression on tissue order.
    
    Parameters:
    -----------
    agg_subset : DataFrame
        Subset of aggregated data for one program and cell type
    covariates : list, optional
        List of covariate column names to include in the model
    
    Returns:
    --------
    dict with test results
    """
    # Filter to valid tissue orders (exclude 999)
    data = agg_subset[agg_subset['tissue_order'] < 999].copy()
    
    if len(data) < 3:  # Need at least 3 tissues
        return {
            'n_tissues': len(data),
            'coef': np.nan,
            'pvalue': np.nan,
            'r_squared': np.nan
        }
    
    # Prepare features
    X = data[['tissue_order']].values
    y = data['mean_score'].values
    
    # Add covariates if provided
    if covariates:
        for cov in covariates:
            if cov in data.columns:
                # Encode categorical covariates
                if data[cov].dtype == 'object':
                    cov_encoded = pd.get_dummies(data[cov], prefix=cov, drop_first=True)
                    X = np.hstack([X, cov_encoded.values])
                else:
                    X = np.hstack([X, data[[cov]].values])
    
    # Fit linear model
    try:
        model = LinearRegression()
        model.fit(X, y)
        y_pred = model.predict(X)
        
        # Calculate R-squared
        ss_res = np.sum((y - y_pred) ** 2)
        ss_tot = np.sum((y - np.mean(y)) ** 2)
        r_squared = 1 - (ss_res / ss_tot) if ss_tot > 0 else 0
        
        # Calculate p-value using F-test
        n = len(y)
        p = X.shape[1]
        if n > p:
            f_stat = (r_squared / p) / ((1 - r_squared) / (n - p - 1))
            pvalue = 1 - stats.f.cdf(f_stat, p, n - p - 1)
        else:
            pvalue = np.nan
        
        return {
            'n_tissues': len(data),
            'coef': model.coef_[0],  # Coefficient for tissue_order
            'pvalue': pvalue,
            'r_squared': r_squared,
            'intercept': model.intercept_
        }
    except:
        return {
            'n_tissues': len(data),
            'coef': np.nan,
            'pvalue': np.nan,
            'r_squared': np.nan
        }
```

```python
# Test segment effects for each program and cell type
# Use celltype_tissue aggregation (sample-level aggregation)
segment_test_results = []

# Get unique combinations
agg_ct_tissue = agg_df[agg_df['aggregation_level'] == 'celltype_tissue'].copy()

for (prog, celltype), group in agg_ct_tissue.groupby(['program', 'Prelim annotation']):
    result = test_segment_effect(group)
    result['program'] = prog
    result['celltype'] = celltype
    segment_test_results.append(result)

segment_tests_df = pd.DataFrame(segment_test_results)

# Calculate FDR
segment_tests_df['fdr'] = multipletests(segment_tests_df['pvalue'], method='fdr_bh')[1]

print(f"Tested {len(segment_tests_df)} program-celltype combinations")
print(f"\nSignificant segment effects (FDR < 0.05): {np.sum(segment_tests_df['fdr'] < 0.05)}")
print(f"\nSample results:")
print(segment_tests_df.head(10))
```

```python
# Save segment test results
segment_tests_df.to_csv(output_dir / 'segment_effect_tests.csv', index=False)
print(f"Saved segment effect tests to: {output_dir / 'segment_effect_tests.csv'}")
```

## Programs Previously Believed Segment-Specific But Found Throughout

This section identifies programs with low tissue specificity by comparing variance across tissues vs across cell types.

```python
# Calculate tissue specificity metrics for each program
tissue_specificity_results = []

for prog_name in program_scores.keys():
    prog_data = agg_df[
        (agg_df['program'] == prog_name) & 
        (agg_df['aggregation_level'] == 'celltype_tissue')
    ].copy()
    
    if len(prog_data) == 0:
        continue
    
    # Calculate variance across tissues (within each cell type)
    tissue_variance = prog_data.groupby('Prelim annotation')['mean_score'].var().mean()
    
    # Calculate variance across cell types (within each tissue)
    celltype_variance = prog_data.groupby('tissue')['mean_score'].var().mean()
    
    # Calculate overall variance
    overall_variance = prog_data['mean_score'].var()
    
    # Calculate coefficient of variation across tissues (mean across cell types)
    tissue_cv = prog_data.groupby('Prelim annotation')['mean_score'].apply(
        lambda x: x.std() / x.mean() if x.mean() != 0 else np.nan
    ).mean()
    
    # Calculate coefficient of variation across cell types (mean across tissues)
    celltype_cv = prog_data.groupby('tissue')['mean_score'].apply(
        lambda x: x.std() / x.mean() if x.mean() != 0 else np.nan
    ).mean()
    
    # Tissue specificity ratio: low ratio = low tissue specificity (high cell type specificity)
    # High ratio = high tissue specificity (low cell type specificity)
    if celltype_variance > 0:
        specificity_ratio = tissue_variance / celltype_variance
    else:
        specificity_ratio = np.nan
    
    tissue_specificity_results.append({
        'program': prog_name,
        'tissue_variance': tissue_variance,
        'celltype_variance': celltype_variance,
        'overall_variance': overall_variance,
        'tissue_cv': tissue_cv,
        'celltype_cv': celltype_cv,
        'specificity_ratio': specificity_ratio
    })

tissue_spec_df = pd.DataFrame(tissue_specificity_results)

# Sort by specificity ratio (low = low tissue specificity, found throughout)
tissue_spec_df = tissue_spec_df.sort_values('specificity_ratio')

print("Tissue Specificity Analysis:")
print("=" * 60)
print("\nPrograms with LOW tissue specificity (found throughout segments):")
print("(Low specificity_ratio = more consistent across tissues)")
print(tissue_spec_df.head(10).to_string(index=False))

print("\n\nPrograms with HIGH tissue specificity (segment-specific):")
print("(High specificity_ratio = varies more across tissues)")
print(tissue_spec_df.tail(10).to_string(index=False))
```

```python
# Visualize tissue specificity
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot 1: Variance comparison
ax1 = axes[0]
ax1.scatter(
    tissue_spec_df['celltype_variance'],
    tissue_spec_df['tissue_variance'],
    s=100,
    alpha=0.6
)
ax1.plot([0, tissue_spec_df[['celltype_variance', 'tissue_variance']].max().max()],
         [0, tissue_spec_df[['celltype_variance', 'tissue_variance']].max().max()],
         'r--', alpha=0.5, label='Equal variance')
ax1.set_xlabel('Variance Across Cell Types', fontsize=12)
ax1.set_ylabel('Variance Across Tissues', fontsize=12)
ax1.set_title('Tissue vs Cell Type Variance', fontsize=14)
ax1.legend()
ax1.grid(True, alpha=0.3)

# Add program labels for extreme points
for idx, row in tissue_spec_df.iterrows():
    if row['specificity_ratio'] < 0.5 or row['specificity_ratio'] > 2:
        ax1.annotate(row['program'], 
                     (row['celltype_variance'], row['tissue_variance']),
                     fontsize=8, alpha=0.7)

# Plot 2: Specificity ratio
ax2 = axes[1]
ax2.barh(range(len(tissue_spec_df)), tissue_spec_df['specificity_ratio'], alpha=0.6)
ax2.set_yticks(range(len(tissue_spec_df)))
ax2.set_yticklabels(tissue_spec_df['program'], fontsize=8)
ax2.set_xlabel('Tissue Specificity Ratio\n(Low = found throughout, High = segment-specific)', fontsize=12)
ax2.set_title('Tissue Specificity by Program', fontsize=14)
ax2.axvline(x=1, color='r', linestyle='--', alpha=0.5, label='Equal variance')
ax2.legend()
ax2.grid(True, alpha=0.3, axis='x')

plt.tight_layout()
fig.savefig(output_dir / 'tissue_specificity_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

print(f"Saved tissue specificity plot to: {output_dir / 'tissue_specificity_analysis.png'}")
```

```python
# Save tissue specificity results
tissue_spec_df.to_csv(output_dir / 'tissue_specificity_metrics.csv', index=False)
print(f"Saved tissue specificity metrics to: {output_dir / 'tissue_specificity_metrics.csv'}")

# Identify programs with low tissue specificity (found throughout)
low_specificity_threshold = 0.5  # Adjust as needed
programs_found_throughout = tissue_spec_df[
    tissue_spec_df['specificity_ratio'] < low_specificity_threshold
]['program'].tolist()

print(f"\nPrograms with low tissue specificity (ratio < {low_specificity_threshold}):")
print("These programs are found throughout gut segments:")
for prog in programs_found_throughout:
    print(f"  - {prog}")
```
